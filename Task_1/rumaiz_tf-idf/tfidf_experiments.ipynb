## 1. Import Libraries

```python
import json
from pathlib import Path
from collections import Counter
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## 2. Load the development dataset

```python
# main directory
data_dir = Path("../../data")

# file path
dev_file = data_dir / "dev_labeled_ChatGPT.jsonl"

dev_data = []

with open(dev_file, "r", encoding="utf-8") as f:
    for line in f:
        if line.strip():
            dev_data.append(json.loads(line))

# just confirming it loads
print(f"Loaded {len(dev_data)} development examples!!")
```

## 3. Load the Wikipedia passages

```python
# passages files
passages_file = data_dir / "passages_bm25_ChatGPT_humfacts.jsonl"

passages_data = []

with open(passages_file, "r", encoding="utf-8") as f:
    for line in f:
        if line.strip():
            passages_data.append(json.loads(line))

print(f"Loaded {len(passages_data)} passage records!!")
```

## 4. Inspect data

```python
print("Development example:")
print(json.dumps(dev_data[0], indent=2, ensure_ascii=False))
print()
print("Passage record:")
print(json.dumps(passages_data[0], indent=2, ensure_ascii=False))
```

```python
print("Development examples:", len(dev_data))
print("Passage records:", len(passages_data))
print()
print("Development record keys:")
print(dev_data[0].keys())
print()
print("Passage record keys:")
print(passages_data[0].keys())
print()
print("Label distribution:")
print(Counter(example.get("label") for example in dev_data))
```

## 5. Fact to passage mapping

```python
# Build fact to passage mapping
fact_to_passage = {}
for rec in passages_data:
    fact_to_passage[rec['sent']] = rec['passages']
```
