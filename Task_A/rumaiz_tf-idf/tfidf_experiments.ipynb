## 5. Fact to passage mapping

```python
# Build fact to passage mapping
fact_to_passage = {}
for rec in passages_data:
    fact_to_passage[rec['sent']] = rec['passages']
```

```python
# build examples maybe
examples = []   # stores facts, passages and labels in a dict

for rec in dev_data:
    # check if annotations exist
    if rec.get('annotations') is None:
        continue
    # iterate through annotations and extract facts and labels
    for sent in rec['annotations']:
        # check if human-atomic-facts exist
        if sent.get('human-atomic-facts') is None:
            continue
        for fact in sent['human-atomic-facts']:
            text = fact['text']
            label = fact['label']
            # Check whether the fact has an associated passage
            if text in fact_to_passage:
                examples.append({
                    'fact': text,
                    'passages': fact_to_passage[text],
                    'label': label
                })
```

```python
print(f"Built {len(examples)} examples")
print(Counter(ex['label'] for ex in examples))
```
